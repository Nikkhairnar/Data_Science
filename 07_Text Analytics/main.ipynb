```python
corpus = ['data science is one of the most important fields of science','this is one of the best data science courses','data scientists analyze data' ]
```

```python
words_set = set()
for doc in corpus:
    words = doc.split(' ')
    words_set = words_set.union(set(words))
print('Number of words in the corpus:',len(words_set))
print('The words in the corpus: \n', words_set)
```

Computing Term Frequency

Now we can create a dataframe by the number of documents in the corpus and the word set,and use that information to compute the term frequency (TF):

```python
import pandas as pd
import numpy as np

n_docs = len(corpus)
n_words_set = len(words_set)

df_tf = pd.DataFrame(
    np.zeros((n_docs, n_words_set)),
    columns=list(words_set)
)

# Compute TF
for i in range(n_docs):
    words = corpus[i].split(' ')
    
    for w in words:
        df_tf.loc[i, w] += (1 / len(words))

df_tf
```

```python
print("IDF of: ")

idf = {}

for w in words_set:
    k = 0 # number of documents in the corpus that contain this word
    for i in range(n_docs):
        if w in corpus[i].split():
            k += 1
            
idf[w] = np.log10(n_docs / k)
print(f'{w:>15}: {idf[w]:>10}' )
```

```python
df_tf_idf = df_tf.copy()

for w in words_set:
    if w in df_tf.columns and w in idf:
        for i in range(n_docs):
            df_tf_idf.loc[i, w] = df_tf.loc[i, w] * idf[w]

df_tf_idf
```
